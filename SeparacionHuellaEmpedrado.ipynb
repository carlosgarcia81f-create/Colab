````python
import pandas as pd
import plotly.graph_objects as go

######----------------------------- B L O Q U E  1 ------------------------------#####
# G R A F I C A D O
# 1. Cargar datos extraídos del levantamiento de campo
data_deterioros = [
    {"Cad_Ini": 8.70,  "Cad_Fin": 50,   "Junta": "J2", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 9.50,  "Cad_Fin": 22.80,  "Junta": "J3", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 9.50,  "Cad_Fin": 22.80,  "Junta": "J4", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 9.50,  "Cad_Fin": 38.40,  "Junta": "J6", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 9.50,  "Cad_Fin": 38.40,  "Junta": "J7", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 28.50, "Cad_Fin": 39.00,  "Junta": "J3", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 59.50, "Cad_Fin": 86.10,  "Junta": "J6", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 62.70, "Cad_Fin": 68.50,  "Junta": "J4", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 109.60,"Cad_Fin": 116.60,  "Junta": "J6", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 162.40,"Cad_Fin": 170.00, "Junta": "J4", "Deterioro": "Separación y Levantado"},
    {"Cad_Ini": 170.90,"Cad_Fin": 193.40, "Junta": "J6", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 200.20,"Cad_Fin": 207.5, "Junta": "J4", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 226.20,"Cad_Fin": 239.00, "Junta": "J3", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 226.20,"Cad_Fin": 239.00, "Junta": "J4", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 264.80,"Cad_Fin": 422.30, "Junta": "J2", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 282.90,"Cad_Fin": 391.00, "Junta": "J3", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 300.00,"Cad_Fin": 352.70, "Junta": "J6", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 307.00,"Cad_Fin": 345.00, "Junta": "J5", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 307.00,"Cad_Fin": 384.90, "Junta": "J4", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 401.40,"Cad_Fin": 422.30, "Junta": "J3", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 401.40,"Cad_Fin": 422.30, "Junta": "J4", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 407.0,"Cad_Fin": 422.30, "Junta": "J5", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 408.40,"Cad_Fin": 430.00, "Junta": "J6", "Deterioro": "Separación huella-empedrado"},
    {"Cad_Ini": 409.80,"Cad_Fin": 427.30, "Junta": "J7", "Deterioro": "Separación huella-empedrado"},
]

df = pd.DataFrame(data_deterioros)
df.to_excel('DataAnalizada.xlsx',
index=False)

# 2. Coordenadas de las juntas según la sección del plano (ancho arroyo = 6.0 m)
posicion_juntas = {
    "J1": 0.00, # Guardalosa izq
    "J2": 0.70, # Huella 1 ext
    "J3": 1.80, # Huella 1 int / empedrado
    "J4": 2.50, # Empedrado / Huella 2 int
    "J5": 2.90, # Eje central
    "J6": 3.60, # Huella 3 int
    "J7": 4.70, # Huella 3 ext / empedrado
    "J8": 5.40  # Guardalosa der
}

fig = go.Figure()

# 3. Dibujar las líneas guías de cada junta longitudinal
max_cad = df["Cad_Fin"].max() + 10
for junta, x_pos in posicion_juntas.items():
    fig.add_shape(
        type="line",
        x0=x_pos, x1=x_pos, y0=0, y1=max_cad,
        line=dict(color="#D1D5DB", width=1.5, dash="dash")
    )

# Palette de colores según severidad o tipo de daño
colores = {
    "Separación huella-empedrado": "#DC2626", # Rojo
    "Separación y Levantado": "#7F1D1D"       # Rojo obscuro
}

# 4. Trazar los tramos afectados con la información flotante (Hover)
for deterioro, grupo in df.groupby("Deterioro"):
    color = colores.get(deterioro, "#F59E0B")
    mostrar_en_leyenda = True

    for _, row in grupo.iterrows():
        x = posicion_juntas[row["Junta"]]
        y0 = row["Cad_Ini"]
        y1 = row["Cad_Fin"] if row["Cad_Fin"] != row["Cad_Ini"] else row["Cad_Ini"] + 0.5
        longitud = round(row["Cad_Fin"] - row["Cad_Ini"], 2)

        fig.add_trace(go.Scatter(
            x=[x, x],
            y=[y0, y1],
            mode="lines+markers",
            line=dict(color=color, width=6),
            marker=dict(size=4),
            name=deterioro,
            legendgroup=deterioro,
            showlegend=mostrar_en_leyenda,
            hovertemplate=(
                f"<b>Junta:</b> {row['Junta']}<br>" +
                f"<b>Tipo:</b> {row['Deterioro']}<br>" +
                f"<b>Tramo:</b> 0+{y0:06.2f} m a 0+{row['Cad_Fin']:06.2f} m<br>" +
                f"<b>Longitud afectada:</b> {longitud} m<extra></extra>"
            )
        ))
        mostrar_en_leyenda = False

# 5. Formato y diseño del gráfico
fig.update_layout(
    title="<b>Levantamiento de Deterioros en Juntas de Huellas (Diagrama de Cadena)</b>",
    xaxis=dict(
        tickmode="array",
        tickvals=list(posicion_juntas.values()),
        ticktext=list(posicion_juntas.keys()),
        title="<b>Juntas Longitud de la Sección Transversal</b>",
        range=[-0.2, 5.6]
    ),
    yaxis=dict(
        title="<b>Cadenamiento (m)</b>",
        #autorange="reversed", # Mantiene 0+000 en la parte superior
        zeroline=True,
        gridcolor="#E5E7EB"
    ),
    height=850,
    width=750,
    template="plotly_white",
    hovermode="closest",
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)

fig.show()


######----------------------------- B L O Q U E  2 ------------------------------#####
# C O N S O L I D A C I Ó N  D E  D E T E R I O R O S
# 1. Mapeo de Juntas a Huellas físicas (según tu detalle de sección transversal)
# Cada Huella "pertenece" a un par de juntas
JUNTA_A_HUELLA = {
    "J1": "Huella 1 (Ext. Izq)",
    "J2": "Huella 1 (Ext. Izq)",
    "J3": "Huella 2 (Int. Izq)",
    "J4": "Huella 2 (Int. Izq)",
    "J5": "Huella 3 (Int. Der)",
    "J6": "Huella 3 (Int. Der)",
    "J7": "Huella 4 (Ext. Der)",
    "J8": "Huella 4 (Ext. Der)"
}

# 2. Datos del levantamiento

df = pd.DataFrame(data_deterioros)

# Función matemática para fusionar intervalos traslapados (Unión de conjuntos de líneas)
def fusionar_intervalos(intervalos):
    if not intervalos:
        return []
    # Ordenar por cadenamiento inicial
    intervalos_ordenados = sorted(intervalos, key=lambda x: x[0])
    fusionados = [list(intervalos_ordenados[0])]

    for inicio, fin in intervalos_ordenados[1:]:
        ultimo_inicio, ultimo_fin = fusionados[-1]
        if inicio <= ultimo_fin: # Existe traslape o continuidad
            fusionados[-1][1] = max(ultimo_fin, fin)
        else:
            fusionados.append([inicio, fin])

    return fusionados

# 3. Procesar datos por Huella
df["Huella"] = df["Junta"].map(JUNTA_A_HUELLA)

resumen_cuantificacion = []

for huella, grupo in df.groupby("Huella"):
    # Obtener pares de (Cad_Ini, Cad_Fin) para esta huella
    intervalos_raw = list(zip(grupo["Cad_Ini"], grupo["Cad_Fin"]))

    # Resolver traslapes en los dos bordes/juntas de la huella
    intervalos_unicos = fusionar_intervalos(intervalos_raw)

    # Calcular longitud sin duplicidades
    longitud_total_huella = sum(fin - inicio for inicio, fin in intervalos_unicos)

    for ini, fin in intervalos_unicos:
        long_tramo = round(fin - ini, 2)
        resumen_cuantificacion.append({
            "Elemento / Huella": huella,
            "Cad_Inicial": f"0+{ini:06.2f}",
            "Cad_Final": f"0+{fin:06.2f}",
            "Longitud Tramo (m)": long_tramo if long_tramo > 0 else 0.50 # Asigna mínimo si es puntual
        })

# 4. Crear DataFrame con el reporte cuantificado
df_resumen = pd.DataFrame(resumen_cuantificacion)

print("--- DETALLE DE TRAMOS A OBSERVAR / REPARAR SIN TRASLAPES ---")
print(df_resumen.to_string(index=False))

print("\n--- RESUMEN TOTAL DE METROS LINEALES POR HUELLA ---")
totales = df_resumen.groupby("Elemento / Huella")["Longitud Tramo (m)"].sum().reset_index()

print(totales.to_string(index=False))
````

````python
import pandas as pd
import numpy as np

# Habilitar la visualización interactiva de DataFrames de forma correcta
%load_ext google.colab.data_table

# ==============================================================================
# 1. MEDIONES DE ANCHO TOTAL DE ARROYO EN CAMPO
# ==============================================================================
mediciones_ancho = [
    {"Cad": 0.00,   "Ancho_Total": 6.10}, #
    {"Cad": 20.00,  "Ancho_Total": 5.45}, #
    {"Cad": 40.00,  "Ancho_Total": 5.46}, #[cite: 3]
    {"Cad": 60.00,  "Ancho_Total": 5.46}, #[cite: 3]
    {"Cad": 89.00,  "Ancho_Total": 5.48}, #[cite: 3]
    {"Cad": 100.40, "Ancho_Total": 5.48}, #[cite: 3]
    {"Cad": 125.00, "Ancho_Total": 5.45}, #[cite: 3]
    {"Cad": 150.00, "Ancho_Total": 5.46}, #[cite: 3]
    {"Cad": 200.00, "Ancho_Total": 5.48}, #[cite: 3]
    {"Cad": 235.00, "Ancho_Total": 5.45}, #[cite: 3]
    {"Cad": 243.00, "Ancho_Total": 5.45}, #[cite: 3]
    {"Cad": 250.00, "Ancho_Total": 5.45}, #[cite: 3]
    {"Cad": 258.40, "Ancho_Total": 5.41}, #[cite: 3]
    {"Cad": 263.10, "Ancho_Total": 5.44}, #[cite: 3]
    {"Cad": 290.00, "Ancho_Total": 5.44}, #[cite: 3]
    {"Cad": 325.00, "Ancho_Total": 5.45}, #[cite: 3]
    {"Cad": 350.00, "Ancho_Total": 5.49}, #[cite: 3]
    {"Cad": 375.00, "Ancho_Total": 5.55}, #[cite: 3]
    {"Cad": 392.40, "Ancho_Total": 5.51}, #[cite: 3]
    {"Cad": 402.90, "Ancho_Total": 5.53}, #[cite: 3]
    {"Cad": 423.60, "Ancho_Total": 5.52}, #[cite: 3]
    {"Cad": 453.80, "Ancho_Total": 5.47}  #[cite: 3]
]

df_anchos = pd.DataFrame(mediciones_ancho)

def obtener_ancho_arroyo_interpolado(cad):
    return np.interp(cad, df_anchos["Cad"], df_anchos["Ancho_Total"])
# ==============================================================================
# 2. JERARQUÍA Y GEOMETRÍA DE FRANJA (DE DERECHA A IZQUIERDA)
# ==============================================================================
# Menor número = Más a la derecha = Mayor jerarquía de absorción
PRIORIDAD_DERECHA_IZQ = {
    "J8": 0,
    "J7": 1, # Más a la derecha
    "J6": 2,
    "J5": 3,
    "J4": 4,
    "J3": 5,
    "J2": 6,
    "J1": 7  # Más a la izquierda
}
# ==============================================================================
# 3. DEFINICIÓN DE FRANJA Y CONTEO DE HUELLAS CONTENIDAS
# ==============================================================================
# Retorna: (Ancho_Franja_Derecha, Num_Huellas_en_Franja)
def obtener_geometria_franja_derecha(junta, cad):
    W_total = obtener_ancho_arroyo_interpolado(cad)
    W_emp_total = max(0, W_total - 2.80) # Empedrado neto variable

    emp_12 = W_emp_total * 0.423
    emp_23 = W_emp_total * 0.154 # Empedrado central entre H2 y H3
    emp_34 = W_emp_total * 0.423

    # Devuelve: (Ancho Franja a la Izq, Num Huellas a la Izq , Ancho Empedrado a la Izq)
    if junta == "J1":   return (0.0, 0, 0.0)
    elif junta == "J2": return (0.70, 1, 0.0)
    elif junta == "J3": return (0.70 + emp_12, 1, emp_12)
    elif junta == "J4": return (0.70*2 + emp_12, 2, emp_12)
    elif junta == "J5": return (0.70*2 + emp_12 + emp_23, 2, emp_12 + emp_23)
    elif junta == "J6": return (0.70*3 + emp_12 + emp_23, 3, emp_12 + emp_23)
    elif junta == "J7": return (0.70*3 + emp_12 + emp_23 + emp_34, 3, emp_12 + emp_23 + emp_34)
    elif junta == "J8": return (0.70*4 + emp_12 + emp_23 + emp_34, 4, emp_12 + emp_23 + emp_34)
    return (0.00, 0, 0.0)

# ==============================================================================
# 4. LEVANTAMIENTO DE DETERIOROS DE CAMPO
# ==============================================================================
#Los deterioros ya fueron capturados previamente en el diccionario data_deterioros

df_det = pd.DataFrame(data_deterioros)

# ==============================================================================
# 5. CUANTIFICACIÓN CON SEPARACIÓN CONCEPTUAL (HUELLA vs EMPEDRADO)
# ==============================================================================
puntos_quiebre = set(df_det["Cad_Ini"]).union(set(df_det["Cad_Fin"])).union(set(df_anchos["Cad"]))
cadenamientos_ordenados = sorted(list(puntos_quiebre))

registros_observacion = []

for i in range(len(cadenamientos_ordenados) - 1):
    c_ini = cadenamientos_ordenados[i]
    c_fin = cadenamientos_ordenados[i+1]
    distancia = round(c_fin - c_ini, 2)

    if distancia <= 0:
        continue

    juntas_activas = df_det[(df_det["Cad_Ini"] <= c_ini) & (df_det["Cad_Fin"] >= c_fin)]["Junta"].unique()

    if len(juntas_activas) > 0:
        # Evaluar qué junta es la dominante (otorga el mayor ancho de franja)
        datos_geometria = [
            (j, *obtener_geometria_franja_derecha(j, c_ini), *obtener_geometria_franja_derecha(j, c_fin))
            for j in juntas_activas
        ]

        # Buscamos la junta más a la derecha y desempacamos los 7 elementos correspondientes
        junta_dom, w_ini, num_h_ini, emp_ini, w_fin, num_h_fin, emp_fin = min(datos_geometria, key=lambda x: PRIORIDAD_DERECHA_IZQ[x[0]])

        w_promedio = (w_ini + w_fin) / 2.0
        num_huellas = num_h_ini # Las huellas en la franja son constantes dentro del sub-tramo

        # Cálculos de Áreas Desglosadas
        area_total = round(distancia * w_promedio, 2) #[cite: 3]
        area_huellas = round(distancia * (num_huellas * 0.70), 2) # 0.70m de ancho nominal por huella[cite: 3]
        area_empedrado = round(area_total - area_huellas, 2)

        registros_observacion.append({
            "Cad_Inicial": f"0+{c_ini:06.2f}",
            "Cad_Final": f"0+{c_fin:06.2f}",
            "Distancia (m)": distancia,
            "Ancho Prom (m)": round(w_promedio, 2),
            "Huellas en Franja": num_huellas,
            "Área Huellas (m²)": area_huellas,
            "Área Empedrado (m²)": area_empedrado,
            "Área Total (m²)": area_total,
            "Junta Dominante": junta_dom
        })

df_resultado = pd.DataFrame(registros_observacion)

# ==============================================================================
# 6. RESUMEN DE AUDITORÍA DESGLOSADO POR CONCEPTO
# ==============================================================================
print("--- CÉDULA DE AUDITORÍA: DESGLOSE DE ÁREAS OBSERVADAS POR CONCEPTO ---")
print(df_resultado.to_string(index=False))

print("\n" + "="*75)
print(" RESUMEN GENERAL DE OBSERVACIÓN DE OBRA:")
print("="*75)
print(f" DISTANCIA TOTAL INTERVENIDA     : {df_resultado['Distancia (m)'].sum():.2f} m")
print(f" ÁREA OBSERVADA EN HUELLAS       : {df_resultado['Área Huellas (m²)'].sum():.2f} m²")
print(f" ÁREA OBSERVADA EN EMPEDRADO     : {df_resultado['Área Empedrado (m²)'].sum():.2f} m²")
print(f" TOTAL SUPERFICIE OBSERVADA      : {df_resultado['Área Total (m²)'].sum():.2f} m²")
print("="*75)

df_resultado.to_excel('resultado.xlsx', index=False)
````

### Función para Graficar por Intervalos Personalizados de Cadenamiento
Definimos una función que filtra los datos dentro del rango de cadenamientos solicitado y genera la visualización correspondiente.

````python
import plotly.graph_objects as go
import pandas as pd

def generar_grafico_intervalo(c_min, c_max):
    # 1. Posición de las juntas en X
    posicion_juntas = {
        "J1": 0.00, "J2": 0.70, "J3": 1.80, "J4": 2.50,
        "J5": 2.90, "J6": 3.60, "J7": 4.70, "J8": 5.40
    }

    # 2. Filtrar el DataFrame según el intervalo de cadenamiento
    # Convertimos los cadenamientos de texto a números flotantes para evaluar
    df_filtrado = df_resultado.copy()
    df_filtrado["c_ini_num"] = df_filtrado["Cad_Inicial"].apply(lambda x: float(x.replace("0+", "")))
    df_filtrado["c_fin_num"] = df_filtrado["Cad_Final"].apply(lambda x: float(x.replace("0+", "")))

    # Seleccionamos las franjas que se traslapen o estén contenidas en el intervalo [c_min, c_max]
    df_segmento = df_filtrado[
        (df_filtrado["c_fin_num"] >= c_min) &
        (df_filtrado["c_ini_num"] <= c_max)
    ].copy()

    fig = go.Figure()

    # 3. Dibujar las líneas guía de fondo acotadas al intervalo
    for junta, x_pos in posicion_juntas.items():
        fig.add_shape(
            type="line",
            x0=x_pos, x1=x_pos, y0=c_min, y1=c_max,
            line=dict(color="#E5E7EB", width=1.5, dash="dash")
        )

    # 4. Dibujar las franjas recortadas/acotadas visualmente al intervalo
    for idx, row in df_segmento.iterrows():
        # Acotamos visualmente el inicio y fin al rango especificado
        c_ini = max(row["c_ini_num"], c_min)
        c_fin = min(row["c_fin_num"], c_max)

        junta_dom = row["Junta Dominante"]
        x_max = posicion_juntas.get(junta_dom, 5.40)

        x_coords = [0, x_max, x_max, 0, 0]
        y_coords = [c_ini, c_ini, c_fin, c_fin, c_ini]

        # Información flotante
        hover_text = (
            f"<b>FRANJA OBSERVADA (Acotada al Tramo)</b><br>"
            f"<b>Junta Dominante:</b> {junta_dom}<br>"
            f"<b>Tramo Original:</b> {row['Cad_Inicial']} a {row['Cad_Final']}<br>"
            f"<b>Distancia Original:</b> {row['Distancia (m)']:.2f} m<br>"
            f"<b>Ancho Promedio:</b> {row['Ancho Prom (m)']:.2f} m<br>"
            f"------------------------------------<br>"
            f"<b>Huellas en Franja:</b> {row['Huellas en Franja']}<br>"
            f"<b>Área Total Franja:</b> {row['Área Total (m²)']:.2f} m²"
        )

        # Trazar polígono de la franja
        fig.add_trace(go.Scatter(
            x=x_coords,
            y=y_coords,
            fill="toself",
            fillcolor="rgba(220, 38, 38, 0.35)",
            line=dict(color="#B91C1C", width=1.5),
            mode="lines",
            hoverinfo="skip",
            showlegend=False
        ))

        # Punto interactivo para el Hover
        x_centro = x_max / 2.0
        y_centro = (c_ini + c_fin) / 2.0

        fig.add_trace(go.Scatter(
            x=[x_centro],
            y=[y_centro],
            mode="markers",
            marker=dict(size=12, color="rgba(0,0,0,0)"),
            name="",
            hovertemplate=hover_text + "<extra></extra>",
            showlegend=False
        ))

    # 5. Formato del Layout adaptado al intervalo
    fig.update_layout(
        title=f"<b>Esquema de Franjas Observadas (Tramo: 0+{c_min:03.0f} a 0+{c_max:03.0f})</b>",
        xaxis=dict(
            tickmode="array",
            tickvals=list(posicion_juntas.values()),
            ticktext=list(posicion_juntas.keys()),
            title="<b>Sección Transversal / Juntas (m)</b>",
            range=[-0.2, 5.6]
        ),
        yaxis=dict(
            title="<b>Cadenamiento (m)</b>",
            range=[c_min, c_max],
            zeroline=True,
            gridcolor="#F3F4F6"
        ),
        height=700,
        width=800,
        template="plotly_white",
        hovermode="closest",
        hoverlabel=dict(
            bgcolor="white",
            font_size=12,
            bordercolor="#B91C1C",
            align="left"
        )
    )

    return fig
````
